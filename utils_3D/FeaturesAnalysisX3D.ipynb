```python
import os
import numpy as np
import pandas as pd
import torch

for ARCH in ["EfficientNetV2-M", "X3D-L"]:
    # Main Headline
    print("=" * 80)
    print(f"Feature Extraction - {ARCH}")
    print("=" * 80)
    for DATASET in ["MultiBypass140"]:#, "VTS", "MultiBypass140", "SAR_RARP50"]: # JIGSAWS
        TASK = ["gestures"] if DATASET != "MultiBypass140" else ["phases", "steps"]
        for t in TASK:
            # ←– POINT THIS AT YOUR FEATURES FOLDER:
            features_dir = f"/home/gabrielg/projects/BoundedFuture++/Bounded_Future_from_GIT/data/{DATASET}/features/{ARCH}/{t}/fold 0"

            rows = []
            for fname in sorted(os.listdir(features_dir))[:]:  # just first 5 for sanity
                path = os.path.join(features_dir, fname)
                try:
                    if fname.endswith(".npy"):
                        feats = np.load(path)
                    else:
                        feats = torch.load(path, map_location="cpu").numpy()
                    rows.append({
                        'video_file': fname,
                        'feat_dim': feats.shape[0],
                        'timesteps': feats.shape[1] if feats.ndim > 1 else np.nan,
                        'mean':       feats.mean(),
                        'std':        feats.std(),
                        'min':        feats.min(),
                        'max':        feats.max(),
                    })
                except Exception as e:
                    rows.append({'video_file': fname, 'error': str(e)})

            df = pd.DataFrame(rows)
            # add header with dataset and task
            header = f"{DATASET} - {t}"
            print(f"\n\n{header}\n{'-' * len(header)}")
            # print the DataFrame in markdown format
            print(df.to_markdown(index=False))
```
